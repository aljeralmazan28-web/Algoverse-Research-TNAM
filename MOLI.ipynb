```python
import pandas as pd, numpy as np, os
d = "/content/data/egfr/EGFR experiments data/"
out = "/content/data/subset/"; os.makedirs(out, exist_ok=True)
rd = lambda f: pd.read_csv(d+f, sep="\t", index_col=0).apply(pd.to_numeric, errors="coerce")

E, M, C = rd("GDSC_exprs.z.EGFRi.tsv"), rd("GDSC_mutations.EGFRi.tsv"), rd("GDSC_CNA.EGFRi.tsv")
R = pd.read_csv(d+"GDSC_response.EGFRi.tsv", sep="\t", index_col=0)
R.index = R.index.astype(str)

# 1. Sex from one score (XIST high = female, Y genes high = male)
markers = [7503, 6192, 8653, 8284, 7404]
score = E.loc[7503] - E.loc[[6192, 8653, 8284, 7404]].mean()
sex = pd.Series(np.where(score > 1, "female", np.where(score < -1, "male", "ambiguous")), index=E.columns)
print("Sex labels (all 940):", sex.value_counts().to_dict())

# 2. Samples present in all four files, with a confident sex label
common = [s for s in E.columns if s in M.columns and s in C.columns and s in R.index and sex[s] != "ambiguous"]
print("Usable samples:", len(common), pd.Series(sex[common]).value_counts().to_dict())

# 3. Balanced random subset: 40 per sex
rng = np.random.RandomState(42)
pick = []
for s in ["female", "male"]:
    ids = [x for x in common if sex[x] == s]
    pick += list(rng.choice(ids, min(40, len(ids)), replace=False))

# 4. Keep the top-1000-variance genes per omics (sex-marker genes excluded from features)
def top(df, n=1000):
    df = df[pick].drop(index=[g for g in markers if g in df.index], errors="ignore")
    return df.loc[df.var(axis=1).nlargest(n).index]

top(E).to_csv(out+"GDSC_exprs.z.EGFRi.tsv", sep="\t")
top(M).to_csv(out+"GDSC_mutations.EGFRi.tsv", sep="\t")
top(C).to_csv(out+"GDSC_CNA.EGFRi.tsv", sep="\t")
Rs = R[R.index.isin(pick)]; Rs.to_csv(out+"GDSC_response.EGFRi.tsv", sep="\t")
sex[pick].to_csv(out+"sex_labels.csv")

print("\nSubset:", len(pick), "samples | response rows:", len(Rs))
print(Rs.groupby("drug")["response"].value_counts().unstack(fill_value=0))
print("Saved to", out)
```
