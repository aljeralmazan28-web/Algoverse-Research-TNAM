```python
import importlib.util
spec=importlib.util.spec_from_file_location('tnam_task1_splits','tnam_task1_splits.py')
m=importlib.util.module_from_spec(spec);spec.loader.exec_module(m)
source_paths={
  'gdsc2':Path('GDSC2_fitted_dose_response_27Oct23.xlsx'),
  'cellosaurus':Path('cellosaurus.txt')
  }
for name,local_path in source_paths.items():
    if local_path.exists():
        print(f'EXISTING {local_path}: {local_path.stat().st_size:,} bytes')
        continue
    try:
        m.download(m.URLS[name],local_path)
        print(f'DOWNLOADED {name}: {local_path.stat().st_size:,} bytes')
    except Exception as exc:
        print(f'NOT DOWNLOADED {name}: {type(exc).__name__}: {exc}')
```

```python
from pathlib import Path
import subprocess,sys
required=[Path('GDSC2_fitted_dose_response_27Oct23.xlsx'),Path('cellosaurus.txt')]
if not all(p.exists() for p in required):
    raise FileNotFoundError('Missing real source files: '+', '.join(str(x) for x in required if not x.exists()))
cmd=[sys.executable,'tnam_task1_splits.py','--mode','real',
     '--crosswalk','inputs/cell_line_crosswalk.csv',
     '--gdsc',str(required[0]),
     '--cellosaurus',str(required[1]),
     '--dataset','GDSC2','--seed','42','--out','final_gdsc2_splits']
if Path('attribute_list.txt.gz').exists():
    cmd.extend(['--harmonizome-list','attribute_list.txt.gz'])
if Path('final_gdsc2_splits/manifest.json').exists():
    print('Existing frozen manifest found. Reusing it without rewriting.')
else:
    subprocess.run(cmd,check=True)
print('Real data partitions:')
for stem in ['unseen_lines','unseen_drugs','unseen_tissues']:
    print('\n'+stem)
    print(pd.read_csv(f'final_gdsc2_splits/{stem}_summary.csv').to_string(index=False))
```

```python
import pandas as pd, json, zipfile
from pathlib import Path
out=Path('final_gdsc2_splits')
for stem,col in [('unseen_lines','cell_line_id'),('unseen_drugs','drug_id'),('unseen_tissues','tissue')]:
    data=pd.read_csv(out/f'{stem}.csv')
    assert data.groupby(col).split.nunique().max()==1, f'LEAKAGE in {stem}'
    assert data.sample_id.is_unique and set(data.split)=={'train','val','test'}
    print(f'{stem}: {len(data):,} observations, {data[col].nunique():,} disjoint groups — PASS')
print('\nSex mapping status:')
print(pd.read_csv(out/'cellosaurus_sex_mapping_audit.csv').groupby(['match_status','reported_donor_sex']).size())
with zipfile.ZipFile('final_gdsc2_splits.zip','w',compression=zipfile.ZIP_DEFLATED) as z:
    for f in sorted(out.glob('*')):
        if f.is_file(): z.write(f,arcname=f.name)
print('Final export ready: final_gdsc2_splits.zip')
try:
    from google.colab import files
    files.download('final_gdsc2_splits.zip')
except ImportError:pass
```
